```python
#value counts for each categoric columns
dfc = ['gender', 'category', 'payment_method', 'shopping_mall']
for col in dfc:
    print(f"Value counts for column '{col}':")
    print(df[col].value_counts())
    print("\n")
```

```python
df['invoice_date'] = pd.to_datetime(df['invoice_date'], format='mixed')
```

```python
df['invoice_month'] = df['invoice_date'].dt.month
df['invoice_year'] = df['invoice_date'].dt.year
```

```python
df['revenue'] = df['price'] * df['quantity']
```

```python
selc = ['category', 'payment_method', 'shopping_mall', 'gender']

for col in selc:
    print(f"Average price for  {col}:")
    print(df.groupby(col)['price'].mean().sort_values(ascending=False))
    print("\n")
```
